```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_excel('Реестр_выполненных_рейсов_за_период.xlsx', header = 1).drop(columns=['Unnamed: 0', 'Unnamed: 13', 'Unnamed: 15'])
df.tail()
```

```python
df.drop(df.index[11268:], inplace = True)
#удаление строк с результатами для отброса
df.tail()
```

```python
x1 = df['Вместимость']
x = [x1]

fig, ax = plt.subplots()
ax.hist(x)
ax.set_title('Распределение вместимости ВС', fontsize=14)
ax.set_ylabel('Количество выполненных рейсов', fontsize=12)
ax.set_xlabel('Вместимость ВС', fontsize=12)
labels = ["Пассажиров"]
plt.xlim(0, 500)
plt.legend(labels=labels)
plt.show();
```

```python
x1 = df['Обслужено пассажиров']
x=[x1]

fig, ax = plt.subplots()
ax.hist(x);
ax.set_title('Распределение обслуженных пассажиров', fontsize=14)
ax.set_ylabel('Количество выполненных рейсов', fontsize=12)
ax.set_xlabel('Обслуженно пассажиров', fontsize=12)
labels = ["Пассажиров"]
plt.legend(labels=labels)
plt.show();
```

```python
x1 = df['Baggage Weight']
x=[x1]

fig, ax = plt.subplots()
ax.hist(x);
ax.set_title('Распределение веса багажа на рейсах', fontsize=14)
ax.set_ylabel('Количество выполненных рейсов', fontsize=12)
ax.set_xlabel('Вес обслуженного багажа', fontsize=12)
plt.xlim(-1000, 7000)
labels = ["КГ"]
plt.legend(labels=labels)
plt.show();
```

```python
x1 = df['МВМ, тонн']
x=[x1]

fig, ax = plt.subplots()
ax.hist(x);
ax.set_title('Распределение максимальной взлетной массы ВС', fontsize=14)
ax.set_ylabel('Количество выполненных рейсов', fontsize=12)
ax.set_xlabel('Максимальная взлетная масса ВС', fontsize=12)
labels = ["тонн"]
plt.legend(labels=labels)
plt.xlim(0, 400)
plt.show();
```

```python
x1 = df['Кл.3']
x=[x1]

fig, ax = plt.subplots()
ax.hist(x);
ax.set_title('Распределение пассажиров экономического класса по количеству выполненных рейсов', fontsize=14)
ax.set_ylabel('Количество выполненных рейсов', fontsize=12)
ax.set_xlabel('Количество пассажиров эконом класса', fontsize=12)
labels = ["пассажиров"]
plt.legend(labels=labels)
plt.xlim(0, 500)
plt.show();
```

```python
sns.boxplot(x=df['Baggage Weight']);
plt.legend(labels=labels)
plt.title('Box Plot распределения веса багажа', fontsize=14)
plt.ylabel('Количество выполненных рейсов', fontsize=12)
plt.xlim(-1000, 7000)
labels = ["КГ"]
```

```python
#Данный график демонстрирует нам распределение среднего веса багажа на рейсах. Нижняя граница среднего веса - 0 кг; верхняя - около 1 700 кг.
#График имеет верхний (около 300 кг) и нижний квартиль (около 900 кг), между которыми находятся 25 и 75 % наших данных.
#Медиана графика проходит в значении около 500 кг
#Правый ус графика более длинный, что говорит о наличии выбросов. Следовательно, в правой части находится больше выбросов, а в левой -
#данные более сконцентрированны. Выбросы графика наблюдаются от 2000 до почти >6000 кг
```

```python
data = df['В/П'].value_counts(normalize=True)

fig, ax = plt.subplots()
ax.pie(data, labels=data.index, autopct='%1.1f%%');
plt.title('Распределение рейсов на вылет и прилет', fontsize=14)
labels = ["% рейсов"]
plt.legend(labels=labels)
```
