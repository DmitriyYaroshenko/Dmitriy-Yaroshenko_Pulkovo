```python
class Chocolate:
  def __init__(self, price: float, filling: str, nut: bool, nut_number: int):
    if price <= 0:
      raise ValueError ("ОШИБКА! Цена батончика должна быть больше нуля!")

    self.price = price
    self.filling = filling
    self.nut = nut
    self.nut_number = nut_number


  def show_my_chocolate (self):
    if self.filling:
      print (f"Шоколадный батончик, добавка: {self.filling}")
    else:
      print ("Обычный шоколадный батончик")

  def count_nuts (self):
    if self.nut == True:
      print ("*" * self.nut_number)

  def __str__(self):
    return f'Шоколадный батончик, {self.price} рублей'
```

```python
try:
  chocolate_1 = Chocolate(4.22, "Нуга", True, 5)
  print(chocolate_1)
  chocolate_1.show_my_chocolate()
  chocolate_1.count_nuts()

  chocolate_2 = Chocolate(-1.21, "Шоколадная паста", True, 3)
  print(chocolate_2)
  chocolate_2.show_my_chocolate()
  chocolate_2.count_nuts()

except ValueError as error:
  print (error)
```
